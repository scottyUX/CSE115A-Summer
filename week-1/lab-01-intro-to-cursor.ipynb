## Step 4 — Planning Before Building

The hardest part of any feature is figuring out **what** to build, not how to build it. Rushing straight to implementation produces code that solves the wrong problem. This step teaches a three-stage planning workflow used in real engineering teams.

```
Grill Me  →  Plan Mode  →  Agent Mode (incremental)
  (What?)       (How?)          (Build)
```

---

### 4.1 Plan Mode vs. Agent Mode

| Mode | Use when | Shortcut |
|---|---|---|
| **Agent Mode** | Quick edits, tasks you've done before, single-file changes | Default |
| **Plan Mode** | Complex features, multiple files, unclear requirements, team review needed | `Shift+Tab` |

Switch between modes using the dropdown in the chat input or `Shift+Tab`.

---

### 4.2 Grill Me — Interview Yourself First

Before opening Plan Mode, use the **Grill Me** technique to surface what you actually want to build. It works like rubber ducking with a relentless interviewer — the agent asks successive questions about your design, resolves decision dependencies, and produces a rich requirements conversation you can hand directly to Plan Mode.

Install the Grill Me skill in Agent chat:

```
npx skills@latest add mattpocock/skills
```

Then invoke it:

```
/grill-me I want to build a grade calculator for CSE 115A students.
```

The agent will interview you — answer every question, even if your answer is "I don't know yet." That uncertainty is exactly what needs to be resolved before planning. A full session takes ~15 minutes and surfaces requirements you wouldn't have thought of on your own.

> **Why this matters:** Plan Mode can only plan what you tell it. Grill Me helps you tell it the right things.

---

### 4.3 Plan Mode — The 5-Step Flow

Once you know what you want to build, switch to Plan Mode (`Shift+Tab`) and describe the feature:

```
Build a grade calculator module in week-1/ with:
- calculate_grade(name, scores) → returns average and letter grade
- validate_scores(scores) → raises ValueError for invalid input
- summarize_class(filepath) → reads a CSV and returns class stats
Follow the python-style rule. Include tests.
```

Plan Mode follows five steps automatically:

1. **Clarifying questions** — the agent asks what it needs to know
2. **Codebase scan** — it reads relevant existing files for context
3. **Plan generation** — a structured TODO list with approach decisions
4. **Review and edit** — you read the plan and adjust it in chat
5. **You trigger the build** — nothing is written until you say go

---

### 4.4 Save the Plan

Before executing, save the plan to your workspace so it's version-controlled and visible to the team:

1. In the Plan Mode output, click **Save to Workspace**
2. This creates `.cursor/plans/` in your repo
3. Commit it: `git add .cursor/plans/ && git commit -m "docs: add grade calculator plan"`

> Plans in `.cursor/plans/` are team assets. They document *why* you built something, not just what you built.

---

### 4.5 Execute Incrementally

This is where most students go wrong: they approve the plan and let the agent build everything at once. Don't.

Run TODO items in small batches:

```
Execute only steps 1 and 2 from the plan. Stop and show me the output.
```

After each batch:
- Read the diff before accepting
- Run any existing tests
- Commit: `git commit -m "feat: add calculate_grade and validate_scores"`
- Then continue to the next batch

**Why small batches?**
- Changes stay readable and reviewable
- Commits are atomic and meaningful
- You can redirect if the implementation drifts from the plan
- Token usage stays controlled

---

### 4.6 When Things Go Wrong — Revert and Re-Plan

If the agent's output isn't what you wanted, **don't patch it**. Patching bad output compounds the problem.

Instead:
1. Revert the changes: `git checkout .`
2. Go back to the plan and refine it — add the missing constraint or clarification
3. Re-execute from the updated plan

> "The hard part is figuring out **what** change should be made. Invest time there, not in fixing broken output." — Cursor docs

## Step 5 — Prompting Basics

The quality of the agent's output is directly proportional to the quality of your prompt. The agent is not psychic — it works with what you give it.

---

### 5.1 The Anatomy of a Good Prompt

Every effective prompt has four elements:

| Element | Purpose | Example |
|---|---|---|
| **Context** | Where are we, what already exists | `In week-1/grade_calculator.py...` |
| **Task** | What exactly needs to happen | `...add a function called validate_score...` |
| **Constraints** | Rules, types, error handling | `...that raises ValueError for scores outside 0–100...` |
| **Output format** | What the result should look like | `...and returns a bool. Include a docstring.` |

---

### 5.2 Weak vs. Strong Prompts

**Weak prompt:**
```
write a grade calculator
```

**Strong prompt:**
```
In week-1/, create a file called grade_calculator.py.
Write a function calculate_grade(name: str, scores: list[float]) -> dict
that returns {"name": name, "average": float, "letter": str}.
Use this grading scale: A=90+, B=80+, C=70+, D=60+, F=below 60.
Raise ValueError if scores is empty or any score is outside 0–100.
Include a docstring.
```

The strong prompt removes every ambiguity. The agent has no decisions to make — only work to do.

---

### 5.3 @ Mentions — Attaching Context

Use `@` to attach specific resources directly to your prompt. This gives the agent precise context instead of making it search the whole codebase.

| @ Mention | What it attaches |
|---|---|
| `@filename.py` | The contents of a specific file |
| `@src/components/` | All files in a folder |
| `@Docs` | Your indexed documentation |
| `@Terminal` | The most recent terminal output |
| `@Commit` | A specific git commit diff |
| `@Branch` | All changes on the current branch |
| `@Browser` | Current browser tab content |
| `@Past Chats` | Context from a previous conversation |

> Use `@` mentions when you know which files are relevant. If you're not sure, skip it — the agent finds relevant files through its own search.

**Example — referencing a file:**
```
@grade_calculator.py has a bug in calculate_grade — it doesn't handle
scores of exactly 0. Fix it and update the docstring.
```

**Example — referencing terminal output:**
```
@Terminal — the test above is failing. Find the root cause and fix it.
```

---

### 5.4 Visual Context

You can attach images directly to your prompt — no manual transcription needed:

- **Drag and drop** a screenshot into the chat input.
- **Paste from clipboard** (`Cmd+V` / `Ctrl+V`) directly into the prompt.

Useful for: pasting a stack trace screenshot, sharing a Figma design mockup, or showing a broken UI to the agent.

---

### 5.5 Model Selection

Different tasks need different models. You can switch mid-conversation:

| Situation | Recommended model |
|---|---|
| Quick edits, simple fixes | Fast model (e.g. cursor-small) |
| Complex reasoning, architecture decisions | Capable model (e.g. Claude Sonnet, GPT-4o) |
| Long context, large codebases | Max Mode |

Click the model name in the chat input to switch. You don't have to start a new conversation.

## Step 6 — Rules

Rules are persistent instructions prepended to the agent's context at the start of every relevant session. The model doesn't retain memory between completions — rules are how you make behavior consistent without re-stating it every time.

> Rules constrain behavior. Skills add capability. Both are version-controlled — neither is a chat message.

---

### 6.1 Types of Rules

| Type | Location | Scope |
|---|---|---|
| **Project Rules** | `.cursor/rules/*.mdc` | This repo only — committed to git, shared with the team |
| **User Rules** | Cursor Settings → Rules | All your projects — personal defaults |
| **Team Rules** | cursor.com org dashboard | All team projects — enforced org-wide |
| **AGENTS.md** | Any directory in the repo | Files in that directory and its children |

---

### 6.2 File Format

Project rules use the `.mdc` extension with YAML frontmatter. Plain `.md` files in `.cursor/rules` are ignored.

```markdown
---
description: "Enforce Python style: type annotations, docstrings, no print debugging"
alwaysApply: false
globs: "**/*.py"
---

Always write Python code with:
- Type annotations on all function signatures.
- Docstrings on all public functions.
- No print statements for debugging — use logging instead.
```

**Key frontmatter fields:**

| Field | Purpose |
|---|---|
| `description` | How the agent decides relevance — be specific |
| `alwaysApply` | `true` = fires every session; `false` = conditional |
| `globs` | File patterns that trigger the rule |

---

### 6.3 Application Modes

How a rule fires determines when students see it in action:

| Mode | When it fires | Best for |
|---|---|---|
| **Always Apply** | Every chat session, no conditions | Universal standards (git commit format, safety rules) |
| **Apply Intelligently** | Agent decides based on `description` | Context-sensitive guidance (auth patterns, API usage) |
| **Apply to Specific Files** | When edited files match `globs` | Language or framework rules (`**/*.py`, `src/**/*.ts`) |
| **Apply Manually** | Only when @-mentioned in chat | Optional reference material, rarely-needed guidance |

---

---

### 6.5 AGENTS.md

`AGENTS.md` is a plain markdown alternative — no frontmatter required. Place it in any directory and it automatically applies to files in that directory and all subdirectories.

```
week-1/
├── AGENTS.md        ← applies only to files in week-1/
├── grade_calculator.py
└── test_grade_calculator.py
```

Use `AGENTS.md` when you want simple, directory-scoped instructions without the `.mdc` format — useful for monorepos where different packages have different conventions.

---

### 6.6 Best Practices

- Keep rules **under 500 lines** — long rules dilute focus.
- **Reference files instead of copying content** — rules that duplicate code go stale.
- Use `globs` to **scope precisely** — a Python rule shouldn't fire on TypeScript files.
- Commit rules to git — they are team standards, not personal preferences.
- If a rule isn't firing, check the `description` — vague descriptions lead to missed invocations.

## Step 7 — Agent Skills

A **skill** is a portable, version-controlled package that teaches the agent how to perform a domain-specific task. Skills live in your repo, load automatically, and work across any agent that supports the Agent Skills standard.

> Skills are different from rules — rules constrain behavior, skills add capability.

---

### 7.1 Project Skills vs. User Skills

| Type | Location | Scope |
|---|---|---|
| **Project** | `.cursor/skills/` or `.agents/skills/` | This repo only — committed to git, shared with the team |
| **User** | `~/.cursor/skills/` | All your projects — personal defaults |

Commit project skills to version control. They are team assets, not personal config.

---

### 7.2 File Format

A skill is a **folder** containing a `SKILL.md` file. The folder name must match the `name` field.

```
.cursor/skills/
└── tdd/
    └── SKILL.md
```

`SKILL.md` uses YAML frontmatter:

```markdown
---
name: tdd
description: Guides test-driven development. Use before implementing any new function.
paths:
  - "**/*.py"
disable-model-invocation: false
---

Always write tests before implementation.

1. Ask the user to describe the expected behavior.
2. Write failing tests that capture that behavior.
3. Only then implement the function to make the tests pass.
4. Run the tests and confirm they are green before committing.
```

**Key fields:**

| Field | Purpose |
|---|---|
| `name` | Skill identifier — must match the folder name |
| `description` | How the agent decides when to apply it — be specific |
| `paths` | Glob patterns — skill only activates for matching files |
| `disable-model-invocation` | `true` = explicit `/skill-name` only; `false` = agent applies automatically |

## Step 8 — Subagents

A **subagent** is a specialized child agent with a focused role, isolated context, and a clear output. The parent agent delegates to it, it does its job, and returns the result.

---

### 8.1 Built-in Subagents

Cursor ships three subagents out of the box — no setup required:

| Subagent | What it does |
|---|---|
| **Explore** | Searches and analyzes your codebase using a fast model |
| **Bash** | Runs shell commands, isolates verbose terminal output |
| **Browser** | Controls a browser via MCP, filters noisy DOM snapshots |

These activate automatically when the agent decides the task fits them.

---

### 8.2 When to Use Subagents

Use a subagent when:
- A task requires **context isolation** (don't pollute the main conversation).
- Work can run **in parallel** (e.g. review + docs at the same time).
- A step needs a **different skill set** than the parent (e.g. security audit vs. feature build).

Don't use subagents for quick, single-step tasks — the startup overhead isn't worth it.

---

### 8.3 Project Subagents vs. User Subagents

| Type | Location | Scope |
|---|---|---|
| **Project** | `.cursor/agents/` in the repo | This project only — version-controlled, shared with the team |
| **User** | `~/.cursor/agents/` | All your projects — personal defaults |

Project subagents take precedence over user subagents. Commit them to git so your whole team benefits.

---

### 8.4 File Format

Each subagent is a markdown file with YAML frontmatter:

```markdown
---
name: qa-agent
description: Writes tests for edge cases and failure paths. Use after Engineer implements a function.
model: inherit
readonly: false
---

You are a QA specialist. Your only job is to write tests.
Focus on failure paths, boundary values, and edge cases — not happy paths.
Always use pytest. Follow the python-style rule.
```

Key fields:

| Field | Purpose |
|---|---|
| `name` | How you invoke it: `/qa-agent` |
| `description` | How the parent decides to delegate — make it specific |
| `model` | `inherit` uses the parent's model |
| `readonly` | Set `true` for agents that should never write files |

## Step 9 — MCP Integration

**Model Context Protocol (MCP)** lets Cursor Agent connect to external tools and data sources — databases, APIs, design tools, deployment platforms, and more.

### 9.1 What is MCP?

MCP is an open standard that defines how AI agents communicate with external tools. Think of it as a USB standard — any MCP-compatible tool plugs into any MCP-compatible agent.

### 9.2 Enable the Course MCP Servers

We will connect three MCP servers used throughout this course: **Vercel** (deployment), **Supabase** (database), and **Figma** (design).

1. Open **Cursor Settings** (`Cmd+,` or `Ctrl+,`).
2. Navigate to **MCP Servers**.
3. Click **+ Add MCP Server**.
4. Paste the following configuration:

```json
{
  "mcpServers": {
    "vercel": {
      "url": "https://mcp.vercel.com",
      "headers": {}
    },
    "supabase": {
      "url": "https://mcp.supabase.com/mcp",
      "headers": {}
    },
    "Figma": {
      "url": "https://mcp.figma.com/mcp",
      "headers": {}
    }
  }
}
```

5. Save settings. Cursor will attempt to connect to each server — a green dot next to each name means it is reachable.

### 9.3 What Each Server Does

| MCP Server | Purpose |
|---|---|
| **Vercel** | Deploy and manage web projects directly from the agent |
| **Supabase** | Query and manage your database, tables, and auth |
| **Figma** | Read design files and extract components, tokens, and specs |
