## Step 4 — Planning Before Building

A key engineering discipline: **always plan before you build**. Cursor Agent can generate a plan before writing any code.

### 4.1 Ask the Agent to Plan

In Agent chat, type:

```
I want to build a Python script that takes a student's name and
a list of scores, then prints their average grade and letter grade.
Before writing any code, give me a step-by-step plan.
```

Read the plan carefully. Ask yourself:
- Does it cover edge cases (empty list, negative scores)?

- Would you change anything before implementation?

## Step 5 — Prompting Basics

The quality of the agent's output depends heavily on the quality of your prompt. A vague prompt produces vague results.

### The Anatomy of a Good Prompt

| Element | Example |
|---|---|
| **Context** | "In this Python project..." |
| **Task** | "...create a function called `calculate_grade`..." |
| **Constraints** | "...that raises a `ValueError` for invalid inputs..." |
| **Output format** | "...and return a dict with `average` and `letter` keys." |

### 5.1 Weak vs. Strong Prompts

**Weak prompt:**
```
write a grade calculator
```

**Strong prompt:**
```
In week-1/, create a file called grade_calculator.py.
Write a function calculate_grade(name: str, scores: list[float]) -> dict
that returns {"name": name, "average": float, "letter": str}.
Use this grading scale: A=90+, B=80+, C=70+, D=60+, F=below 60.
Raise ValueError if scores is empty or any score is outside 0–100.
Include a docstring.
```

## Step 6 — Rules

Rules are persistent instructions that shape how the agent behaves across every session. Think of them as standing orders.

### Types of Rules in Cursor

| Type | Location | Scope |
|---|---|---|
| **Project Rules** | `.cursor/rules/*.mdc` | This repo only, version-controlled |
| **User Rules** | Cursor Settings | All your projects |
| **Team Rules** | Org dashboard | All team projects |
| **AGENTS.md** | Repo root | Portable, plain markdown |

### 6.1 Create Your First Project Rule

In Agent chat, type:

```
/create-rule
```

When prompted, name the rule `python-style` and give it these instructions:

```
Always write Python code with:
- Type annotations on all function signatures.
- Docstrings on all public functions.
- No print statements for debugging — use logging instead.
```

### 6.2 Verify the Rule is Applied

Ask the agent to write a new small function and confirm it follows the rule automatically without you re-stating it.

### ✅ Checkpoint 4
A `.cursor/rules/python-style.mdc` file should exist and the agent's next code output should include type annotations and docstrings.

## Step 7 — Agent Skills

Skills are reusable, composable agent capabilities — pre-packaged instructions the agent can invoke by name.

### 7.1 What Makes a Skill

A skill has three parts:
- **Name** — how you invoke it (e.g. `code-review`)
- **Scope** — what files or contexts it applies to
- **Success criteria** — how you know the skill did its job

## Step 8 — Subagents

A **subagent** is a specialized child agent that a parent agent delegates work to. This is how complex tasks get broken into manageable pieces.

### 8.1 How Subagents Work

```
Parent Agent
  ├── Subagent A: "Write the tests"
  ├── Subagent B: "Implement the function"
  └── Subagent C: "Write the documentation"
```

Each subagent gets a specific task, runs independently, and returns results to the parent.

### 8.2 Trigger a Multi-Step Task

In Agent chat, give the agent a task that requires planning and delegation:

```
I need to add a new feature to grade_calculator.py: a function
that reads a CSV file of student scores and prints a class summary
(average, highest, lowest, grade distribution).

Plan the work, implement it, and write tests for it.
Do each step separately and show me before proceeding.
```

Watch how the agent breaks this into discrete steps.

## Step 9 — MCP Integration

**Model Context Protocol (MCP)** lets Cursor Agent connect to external tools and data sources — databases, APIs, browsers, file systems, and more.

### 9.1 What is MCP?

MCP is an open standard that defines how AI agents communicate with external tools. Think of it as a USB standard — any MCP-compatible tool plugs into any MCP-compatible agent.

### 9.2 Enable an MCP Server

1. Open **Cursor Settings** (`Cmd+,` or `Ctrl+,`).
2. Navigate to **MCP Servers**.
3. Click **+ Add MCP Server**.
4. Add the following configuration for a filesystem server:

```json
{
  "mcpServers": {
    "filesystem": {
      "command": "npx",
      "args": ["-y", "@modelcontextprotocol/server-filesystem", "/path/to/CSE115A-Summer"]
    }
  }
}
```

> Replace `/path/to/CSE115A-Summer` with the actual path on your machine.

### 9.3 Confirm the Tool is Reachable

In Agent chat:

```
Using the filesystem MCP tool, list all files in the week-1 folder.
```

### ✅ Checkpoint 7
The agent should list files in `week-1/` using the MCP tool, not its built-in file reader.
