# Lab 03 — Cursor Rules

**Time:** 10 minutes

Write one project rule in [repo-analysis-sprint](https://github.com/scottyUX/repo-analysis-sprint) and open a pull request.

---

## Rules Intro

Cursor **rules** are persistent instructions included in the agent's context. They are version-controlled team standards — not one-off chat messages.

Full reference: [cursor.com/docs/rules](https://cursor.com/docs/rules)

---

### Where project rules live

- Path: `.cursor/rules/*.mdc`
- Must use the **`.mdc`** extension — plain `.md` files in this folder are **ignored**

---

### File anatomy

Every rule has **YAML frontmatter** + a **markdown body**:

```markdown
---
description: "TypeScript strict typing for all .ts files"
alwaysApply: false
globs: "**/*.ts"
---

# TypeScript Standards
- Never use `any`.
- Annotate all function parameters and return types.
```

**Three frontmatter fields** (match the type dropdown in Cursor Settings):

| Field | Purpose |
|---|---|
| `description` | Short summary — used for intelligent matching |
| `globs` | File patterns that trigger the rule |
| `alwaysApply` | If `true`, rule loads in every session |

---

### Four rule types

| Rule type | Settings | When it applies |
|---|---|---|
| **Always Apply** | `alwaysApply: true` | Every chat session |
| **Apply to Specific Files** | `alwaysApply: false` + `globs` | When a matching file is in context |
| **Apply Intelligently** | `alwaysApply: false` + `description` (no globs) | Agent decides from description |
| **Apply Manually** | no globs, no description | Only when `@`-mentioned in chat |

**Behavior cheat sheet:**

| `alwaysApply` | `description` | `globs` | Result |
|---|---|---|---|
| `true` | — | — | Always included |
| `false` | — | set | Fires when glob matches |
| `false` | set | — | Agent pulls in when relevant |
| `false` | — | — | Manual `@` only |

**Glob examples:** `**/*.ts` · `src/components/**/*.tsx`

---

### Rule scope (by location)

| Type | Location | Who shares it |
|---|---|---|
| **Project** | `.cursor/rules/` in the repo | Whole team via git |
| **User** | Cursor Settings → Rules | You only |
| **Team** | Org dashboard (paid plans) | Whole organization |

This lab writes **project rules** only.

---

### Writing good rules

- Keep each rule **short** — one concern per file
- Be **specific** in `description` (especially for Apply Intelligently)
- Use **globs** so rules do not fire on unrelated files
- Use **Always Apply sparingly** — it loads every session

---

### AGENTS.md

Plain markdown alternative in the project root or a subfolder. No frontmatter or globs — applies to that directory and its children. See the [Cursor Rules docs](https://cursor.com/docs/rules) for details.

```markdown
---
description: "Enforce strict TypeScript typing across all .ts files"
alwaysApply: false
globs: "**/*.ts"
---

# TypeScript Strict Standards

- Never use `any` or `unknown` for metrics or API data.
- Annotate every function parameter and return type.
- Use explicit `interface` or `type` for nested data structures.
- If a type is missing, add it before implementing logic.
```

<details>
<summary><strong>Team 3 — Apply Intelligently</strong></summary>

```markdown
---
description: "Resilient async fetch and error handling for repository analysis dashboard code. Use when editing fetch flows, loading states, or error UI in repo-analysis-dashboard.tsx or repo-analysis.utils.ts."
alwaysApply: false
---

# Async Error Handling

- Every async fetch path must use `try/catch` or explicit `.catch()`.
- Store failures in React error state — never crash the UI silently.
- Show the Connection Failed warning card when fetch fails.
- Clear stale errors before starting a new request.
- Reset loading state in a `finally` block.
```

</details>

<details>
<summary><strong>Team 4 — Rule with good/bad examples</strong></summary>

```markdown
---
description: "React component patterns for repository analysis dashboard UI"
alwaysApply: false
globs: "src/components/**/*.tsx"
---

# Component Patterns

Render lists with stable keys and visible empty states.

```tsx
// BAD — index key, no empty state
{items.map((item, i) => <span key={i}>{item}</span>)}

// GOOD — stable key, empty fallback
{items.length === 0
  ? <p>No items found.</p>
  : items.map((item) => <Badge key={item.id}>{item.label}</Badge>)}
```

- Map over arrays with a stable, readable `key`.
- Handle empty lists gracefully in the UI.
- Keep rendering helpers small; add JSDoc when logic is non-obvious.
```

</details>

Verify: **Cursor Settings → Rules** — your file should appear.

---

### Step 4 — Open a PR (2 min)

```bash
git checkout -b team-N-rules-lab
git add .cursor/rules/team-N-*.mdc
git commit -m "docs: add team N cursor rule (Lab 03)"
git push -u origin team-N-rules-lab
```
