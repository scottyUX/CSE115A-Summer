## Spec-Driven Development Intro

**The problem:** Jumping straight to code ("vibe coding") skips requirements. The agent builds the wrong thing, and you rework it later.

**The fix:** Spec-Driven Development (SDD) treats specifications as **executable artifacts** — markdown files the agent reads and acts on, not documents you throw away after planning.

Full methodology: [spec-driven.md](https://github.com/github/spec-kit/blob/main/spec-driven.md)

---

### Core lean loop

```
Constitution  →  Specify  →  Plan  →  Tasks  →  Implement
 (principles)     (what/why)  (how)   (work)     (code)
```

**Artifact chain:**

| Step | Command | Output file |
|---|---|---|
| Principles | `/speckit.constitution` | `.specify/memory/constitution.md` |
| Requirements | `/speckit.specify` | `specs/<feature>/spec.md` |
| Architecture | `/speckit.plan` | `specs/<feature>/plan.md` |
| Work breakdown | `/speckit.tasks` | `specs/<feature>/tasks.md` |
| Build | `/speckit.implement` | Code + tests in the repo |

Each file feeds the next step — the agent gets **structured context** instead of one vague prompt.

---

### Rules vs Spec Kit

| Tool | Role |
|---|---|
| **Cursor rules** (`.mdc`) | Standing orders — how the agent behaves on every task |
| **Spec Kit** | Feature workflow — what to build, step by step, for one feature |

---

### Optional quality gates (production teams)

For real projects, teams also use `/speckit.clarify`, `/speckit.analyze`, `/speckit.checklist`, and `/speckit.converge`. See the [Quick Start](https://github.github.io/spec-kit/quickstart.html) — this lab uses the **lean path** only.


## Setup — 5 Minutes

### Prerequisites

| Tool | Purpose |
|---|---|

### Step 1 — Install Spec Kit CLI

```bash
uv tool install specify-cli --from git+https://github.com/github/spec-kit.git@v0.12.4
specify --version
```

---

### Step 2 — Clone and initialize

This repo already has code. Spec Kit adds `.specify/` templates and Cursor slash commands on top of it (brownfield setup).

```bash
git clone https://github.com/scottyUX/repo-analysis-sprint.git
cd repo-analysis-sprint
specify init . --integration cursor-agent --force
cursor .
```

Integration details: [cursor-agent](https://github.github.io/spec-kit/reference/integrations.html)

---

### Step 3 — Verify slash commands

Open Cursor Agent chat. You should see these commands available:

- `/speckit.constitution`
- `/speckit.specify`
- `/speckit.plan`
- `/speckit.tasks`
- `/speckit.implement`

If they are missing, re-run `specify init . --integration cursor-agent --force` and restart Cursor.


## Exercise — Lean SDD Workflow (~25 min)

Run these slash commands **in order** in Cursor Agent chat.

---

### Step 1 — Constitution

```
/speckit.constitution Create principles focused on: TypeScript strict typing (no any), unit tests with Vitest, run npm run lint and npm test before any PR, and clear error handling in the dashboard UI.
```

This writes `.specify/memory/constitution.md` — project principles the agent follows in every later step.

---

### Step 2 — Specify (what & why)

Paste this user story after `/speckit.specify`:

```
Add a risk level badge to the repository analysis dashboard. Show Low (green), Medium (yellow), or High (red) based on calculateRiskLevel() thresholds. Handle empty or missing metrics gracefully. Include unit tests for all three risk levels.
```

**Important:** Focus on *what* and *why* — do not mention tech stack here. Spec Kit creates `specs/<feature>/spec.md` and a feature branch.

---

### Step 3 — Plan (how)

```
/speckit.plan Use the existing repo stack: Next.js, TypeScript, Vitest. Extend src/components/repo-analysis-dashboard.tsx and src/lib/repo-analysis.utils.ts. Follow patterns already in the codebase.
```

Review the generated `specs/<feature>/plan.md` before continuing.

---

### Step 4 — Tasks

```
/speckit.tasks
```

Open `specs/<feature>/tasks.md`. Confirm tasks are ordered, name specific files, and include tests.

---

### Step 5 — Implement

```
/speckit.implement
```

When the agent finishes, verify locally:

```bash
npm run lint
npm test
```

Fix any failures in Agent chat, then re-run tests.

---

### Step 6 — Open a PR

```bash
git push -u origin HEAD
```

Open a PR to **`scottyUX/repo-analysis-sprint`** → **`main`**.

**PR title:** `Lab 03 SDD: risk level badge`

**PR body must include:**
- Link to `specs/<feature>/spec.md`
- Verification: `npm run lint` and `npm test` output


## Artifact Tour

After `/speckit.specify`, inspect what Spec Kit created:

```
.specify/
  memory/
    constitution.md       ← project principles (Step 1)
  templates/              ← Spec Kit templates (do not edit in lab)
specs/
  00X-feature-name/
    spec.md               ← what & why (Step 2)
    plan.md               ← how — stack & architecture (Step 3)
    tasks.md              ← ordered, file-specific work (Step 4)
```

**Course connection:** This maps to the agentic workflow you have been using all quarter:

```
Understand codebase  →  Plan  →  TDD  →  Verify  →  Commit
```

Spec Kit automates the **artifact chain** between those steps. Your Cursor rules (from the rules sprint) still govern *how* the agent behaves while it executes the spec.
