## Step 1 — Why Guardrails?

Guardrails are **hard controls** on what the agent can do. Rules are **soft instructions** on how it should behave.

In Lab 01 you created rules — persistent text the agent reads at the start of a session. Rules work well for style, conventions, and workflow expectations. But the agent can still ignore a rule. Guardrails enforce limits at the tool level.

### Rules vs. Guardrails

| | Rules | Guardrails |
|---|---|---|
| **Mechanism** | Text prepended to the agent's context | Approvals, mode locks, hooks |
| **Compliance** | Probabilistic — the agent is instructed to follow | Enforced — the action is blocked or requires your approval |
| **Where it lives** | `.cursor/rules/`, `AGENTS.md`, Cursor Settings | `.cursor/hooks.json`, Cursor Settings, Agent modes |
| **Primary goal** | Consistency and team standards | Safety and blast-radius control |

### Agent Architecture

Week 3 frames a complete agent system as five elements:

```
Agent = Model + Tools + Instructions + Memory + Guardrails
```

Rules sit under **Instructions**. Guardrails are a separate pillar.

> Rules ask the agent to behave. Guardrails stop it when it doesn't.

## Step 2 — Three Types of Guardrails

The course defines three guardrail types. Each type maps to one Cursor mechanism.

| Type | What it prevents | Cursor example |
|---|---|---|
| **Approval gates** | Actions running without your review | **Cursor Settings → Agents → enable approval before terminal commands** — when the agent tries to run a shell command, you must approve it first |
| **File locks** | Agent reading or modifying protected paths | **`beforeReadFile` hook** that denies reads of `.env*` files (e.g. `.env.local` in a Next.js project) |
| **Execution restrictions** | Agent using tools or commands it shouldn't | **Ask mode** — the agent can search and explain but cannot edit files or run destructive commands |

### Same Policy, Two Layers

You can express the same policy as both a rule and a guardrail:

- **Soft:** an `alwaysApply` rule — "Never read `.env` files"
- **Hard:** a `beforeReadFile` hook that denies the read

The rule asks the agent to comply. The hook enforces it even if the agent tries anyway.

## Step 3 — Exercise: Block Destructive Shell Commands

In this exercise you will create a **project hook** — a guardrail you can commit to git and share with your team. This is an **execution restriction**: it blocks specific shell commands before they run.

1. **Open Cursor** on your course repo (or your Next.js project repo).

2. **Create the guardrail.** In Agent chat, paste:

```
Create a project guardrail that blocks destructive shell commands.

Add .cursor/hooks.json with a beforeShellExecution hook.
Add .cursor/hooks/block-destructive.sh that denies:
- git push with --force
- rm -rf

Set failClosed: true on the hook.
Return a clear user_message when a command is blocked.
Make the script executable.
```

Review the generated files before accepting. You should end up with:

- `.cursor/hooks.json`
- `.cursor/hooks/block-destructive.sh`

3. **Verify the guardrail.** In Agent mode, send:

```
Run: git push --force origin main
```

You should see the hook **block or require approval** — the command must not run silently.

4. **Commit your work.** In the Cursor terminal:

```
git add .cursor/hooks.json .cursor/hooks/
git commit -m "feat: add guardrail hook for destructive shell commands"
```
